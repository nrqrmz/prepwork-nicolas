## 1. ¿Qué es pandas?

**pandas** es la librería más usada en Python para **manipular datos en forma de tabla**.

Imagina una hoja de cálculo de Excel: tiene **filas** (registros) y **columnas** (variables). pandas te deja trabajar con esas tablas, pero **programáticamente**: puedes filtrarlas, agruparlas, combinarlas y analizarlas con código.

### Las dos estructuras fundamentales de pandas

| Estructura | Analogía | ¿Qué es? |
|---|---|---|
| **`Series`** | Una **columna** de Excel | Una secuencia de valores con un índice (etiqueta para cada fila). |
| **`DataFrame`** | La **hoja completa** de Excel | Una tabla con múltiples columnas (cada columna es una `Series`). |

En las próximas secciones vamos a crear las dos. 👇

## 6. Acceder a una o varias columnas

Las columnas se acceden con **corchetes** y el nombre entre comillas:

- `df["col"]` → devuelve una **`Series`** (una sola columna).
- `df[["col1", "col2"]]` → devuelve un **`DataFrame`** (varias columnas). Nota los **dobles corchetes**.

### Demo

```python
# Single brackets → Series
island_series = df["island"]
print(type(island_series))
island_series.head()
```

```python
# Double brackets → DataFrame
two_columns_df = df[["island", "body_mass_g"]]
print(type(two_columns_df))
two_columns_df.head()
```

## 7. Acceder a filas con `.iloc`

Para acceder a **filas** la forma más directa es **`.iloc[posición]`** (i de *integer location* — selecciona por **número de posición**, empezando en 0).

- `df.iloc[0]` → primera fila (devuelve una `Series`).
- `df.iloc[10]` → fila en la posición 10.
- `df.iloc[0:5]` → las primeras 5 filas (devuelve un `DataFrame`).

👉 En el próximo notebook veremos `.loc[]`, que selecciona por **etiqueta** del índice en lugar de por posición.

### Demo

```python
# First row of the DataFrame
df.iloc[0]
```

```python
# First 5 rows (same as df.head())
df.iloc[0:5]
```
